## Concept → explain it yourself
Segmentation predicts a label for each pixel. **Semantic segmentation** gives pixels a class; separate objects of the same class can share that class label. **Instance segmentation** gives each object a separate mask and class. **Panoptic segmentation** combines semantic labels for background/stuff and distinct instance labels for countable things.

A **mask** follows the object boundary. A **bounding box** is a coarse rectangle; it is compact and convenient for detection but loses shape detail and may include background. Applications include medical boundaries, road scenes, agriculture, retail, and manufacturing defects. Pixel accuracy can hide weak small-object performance; IoU and Dice measure overlap.

## Simple synthetic example
Build one semantic label map and two object instance IDs. Every pixel belongs to one semantic class, while instance IDs distinguish the two objects.

```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
height,width=96,128
y_grid,x_grid=np.ogrid[:height,:width]
semantic=np.zeros((height,width),dtype=np.uint8)
instance=np.zeros((height,width),dtype=np.uint8)
circle=(x_grid-38)**2+(y_grid-45)**2<22**2
rectangle=(x_grid>=69)&(x_grid<111)&(y_grid>=25)&(y_grid<72)
semantic[circle]=1; semantic[rectangle]=1
instance[circle]=1; instance[rectangle]=2
print('shape:',semantic.shape,'semantic classes:',np.unique(semantic),'instance IDs:',np.unique(instance))
fig,axes=plt.subplots(1,2,figsize=(9,3.5))
axes[0].imshow(semantic,cmap='viridis',vmin=0,vmax=2); axes[0].set_title('Semantic mask')
axes[1].imshow(instance,cmap='tab10',vmin=0,vmax=2); axes[1].set_title('Instance mask')
for ax in axes: ax.axis('off')
plt.tight_layout(); plt.show()
```

```python
def box_from_mask(mask):
    rows,cols=np.where(mask)
    if len(rows)==0: return None
    return int(cols.min()),int(rows.min()),int(cols.max()+1),int(rows.max()+1)
boxes=[box_from_mask(instance==object_id) for object_id in (1,2)]
print('boxes (x_min,y_min,x_max,y_max):',boxes)
fig,ax=plt.subplots(figsize=(5,4)); ax.imshow(instance,cmap='tab10',vmin=0,vmax=2)
for x0,y0,x1,y1 in boxes: ax.add_patch(Rectangle((x0,y0),x1-x0,y1-y0,fill=False,edgecolor='red',linewidth=2))
ax.set_title('Boxes enclose objects; masks follow shape'); ax.axis('off'); plt.show()
def iou(prediction,target):
    intersection=np.logical_and(prediction,target).sum(); union=np.logical_or(prediction,target).sum()
    return float(intersection/union) if union else 1.0
print('self IoU:',iou(instance==1,instance==1))
```

## Architecture ideas and interpretation
- **FCN:** replaces dense layers with convolutional output maps and upsamples to image resolution. A straightforward semantic-segmentation design; coarse maps can blur boundaries.
- **U-Net:** encoder/decoder with skip connections that restore spatial detail. Common when precise boundaries and limited labeled data matter, including biomedical images.
- **Mask R-CNN:** detects proposals and predicts class, box, and per-instance mask. Flexible for instance segmentation, but heavier than a semantic model.
