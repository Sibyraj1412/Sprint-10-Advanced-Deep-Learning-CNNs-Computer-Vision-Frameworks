## Dataset selection and validation
Select a dataset independently; verify its license and label definitions. Split by source entity (patient, person, product, video, field, or machine) to prevent leakage.

Classification folder layout: `dataset/train/class_name/image.jpg`, `dataset/val/class_name/image.jpg`, `dataset/test/class_name/image.jpg`. For segmentation, document aligned image/mask IDs; for detection, document box coordinate order and normalization convention.

```python
from pathlib import Path
from collections import Counter
from PIL import Image,UnidentifiedImageError
from torchvision import datasets,transforms
from torch.utils.data import DataLoader
DATA_ROOT=Path('your_dataset')
image_extensions={'.jpg','.jpeg','.png','.bmp','.webp'}
if not DATA_ROOT.exists():
    print('Choose a dataset, set DATA_ROOT, and rerun. No challenge dataset is assumed.')
else:
    paths=[p for p in DATA_ROOT.rglob('*') if p.suffix.lower() in image_extensions]
    valid=[]; corrupt=[]; sizes=[]
    for path in paths:
        try:
            with Image.open(path) as image: image.verify()
            with Image.open(path) as image: sizes.append(image.size)
            valid.append(path)
        except (UnidentifiedImageError,OSError,ValueError): corrupt.append(path)
    print('files:',len(paths),'valid:',len(valid),'corrupt:',len(corrupt),'examples:',corrupt[:5])
    print('common image sizes:',Counter(sizes).most_common(10))
    train_dir=DATA_ROOT/'train'; val_dir=DATA_ROOT/'val'
    if train_dir.exists() and val_dir.exists():
        transform=transforms.Compose([transforms.Resize((128,128)),transforms.ToTensor()])
        train_data=datasets.ImageFolder(train_dir,transform=transform); val_data=datasets.ImageFolder(val_dir,transform=transform)
        assert train_data.class_to_idx==val_data.class_to_idx,'Train/validation class mapping differs'
        print('mapping:',train_data.class_to_idx,'counts:',Counter(train_data.targets))
        train_loader=DataLoader(train_data,batch_size=32,shuffle=True); val_loader=DataLoader(val_data,batch_size=64,shuffle=False)
    else: print('Organize classification images into train/ and val/ class folders or adapt for masks/boxes.')
```
