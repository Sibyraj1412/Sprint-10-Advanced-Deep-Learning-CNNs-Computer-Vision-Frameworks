## Architecture comparison
| Family | Core idea / innovation | Advantages | Limitations and typical use |
|---|---|---|---|
| AlexNet | Deep CNN, ReLU, GPU training, augmentation popularized at scale | Historically influential; teaching baseline | Large by modern standards; historical comparisons |
| VGG | Repeated small 3×3 convolutions | Uniform, interpretable stack | High parameter/memory cost; feature extraction |
| ResNet | Residual skip connections learn refinements and ease deep optimization | Strong general-purpose pretrained backbone | Compute intensive at high depth/resolution |
| DenseNet | Dense connections reuse preceding feature maps | Feature reuse and gradient flow | Concatenation costs activation memory |
| MobileNet | Depthwise-separable convolutions | Efficient mobile/edge inference | Hardware-dependent accuracy/latency trade-off |
| EfficientNet | Compound scaling balances depth, width, resolution | Accuracy/compute balance | Scaling choices and deployment sizes |

In your own words, state the engineering problem each architecture's main idea addresses.

```python
from torchvision import models
constructors={'AlexNet':models.alexnet,'VGG11':models.vgg11,'ResNet18':models.resnet18,'DenseNet121':models.densenet121,'MobileNetV3-small':models.mobilenet_v3_small,'EfficientNet-B0':models.efficientnet_b0}
for name,make in constructors.items():
 model=make(weights=None); print(name,'parameters:',f'{sum(p.numel() for p in model.parameters()):,}')
```

## Interpret the result
These are parameter counts for random models, not accuracy measurements. Parameter count is not latency: memory bandwidth, batch size, resolution, accelerator, and operator support matter. VGG is easy to study; ResNet is a robust baseline; MobileNet targets constrained deployment. Benchmark the real inference path on target hardware before choosing.
