# Notebook 20 – Model Comparison Report



```python
import io, time, copy
import numpy as np
import torch
from torch import nn
torch.manual_seed(42); np.random.seed(42)
device=torch.device('cuda' if torch.cuda.is_available() else 'cpu')
print('PyTorch:',torch.__version__,'| device:',device)
```

## Comparison protocol
The source task for transfer is even/odd parity learned from training examples; its convolutional encoder initializes the ten-class target task. This is an offline task-transfer demonstration, not ImageNet transfer. All approaches share one stratified test split. Do not select from test metrics.

```python
from sklearn.datasets import load_digits
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score,precision_recall_fscore_support
from torch.utils.data import DataLoader,TensorDataset
import torch.nn.functional as F
d=load_digits(); images=d.images.astype(np.float32)/16; labels=d.target.astype(np.int64)
train_idx,rest=train_test_split(np.arange(len(labels)),test_size=.3,random_state=21,stratify=labels)
val_idx,test_idx=train_test_split(rest,test_size=.5,random_state=21,stratify=labels[rest])
x=torch.tensor(images,dtype=torch.float32).unsqueeze(1); x=F.interpolate(x,(16,16),mode='bilinear',align_corners=False); y=torch.tensor(labels,dtype=torch.long)
train_loader=DataLoader(TensorDataset(x[train_idx],y[train_idx]),batch_size=128,shuffle=True)
parity_loader=DataLoader(TensorDataset(x[train_idx],y[train_idx]%2),batch_size=128,shuffle=True)
test_loader=DataLoader(TensorDataset(x[test_idx],y[test_idx]),batch_size=128)
rf=RandomForestClassifier(n_estimators=200,random_state=21,n_jobs=-1); start=time.perf_counter(); rf.fit(images[train_idx].reshape(len(train_idx),-1),labels[train_idx]); rf_seconds=time.perf_counter()-start
class CompareCNN(nn.Module):
    def __init__(self,n_classes=10):
        super().__init__(); self.features=nn.Sequential(nn.Conv2d(1,16,3,padding=1),nn.ReLU(),nn.MaxPool2d(2),nn.Conv2d(16,32,3,padding=1),nn.ReLU(),nn.AdaptiveAvgPool2d((2,2))); self.head=nn.Linear(128,n_classes)
    def forward(self,x): return self.head(self.features(x).flatten(1))
def train_model(model,loader,epochs=3):
    model=model.to(device); opt=torch.optim.Adam(model.parameters(),lr=1e-3); start=time.perf_counter()
    for _ in range(epochs):
        model.train()
        for xb,yb in loader:
            xb,yb=xb.to(device),yb.to(device); opt.zero_grad(); loss=nn.functional.cross_entropy(model(xb),yb); loss.backward(); opt.step()
    return model,time.perf_counter()-start
scratch,scratch_seconds=train_model(CompareCNN(),train_loader)
source,source_seconds=train_model(CompareCNN(2),parity_loader,epochs=2)
transfer=CompareCNN(); transfer.features.load_state_dict(source.features.state_dict()); transfer,transfer_seconds=train_model(transfer,train_loader)
```

```python
@torch.no_grad()
def predictions(model,loader):
    model.eval(); pred=[]; truth=[]
    for xb,yb in loader: pred.extend(model(xb.to(device)).argmax(1).cpu().tolist()); truth.extend(yb.tolist())
    return np.asarray(pred),np.asarray(truth)
def metrics(pred,truth):
    p,r,f,_=precision_recall_fscore_support(truth,pred,average='macro',zero_division=0)
    return accuracy_score(truth,pred),p,r,f
def model_latency(model,batch,repeats=40):
    model.eval(); batch=batch.to(next(model.parameters()).device)
    with torch.no_grad():
        for _ in range(5): model(batch)
        start=time.perf_counter()
        for _ in range(repeats): model(batch)
    return (time.perf_counter()-start)*1000/repeats
def serialized_size(obj):
    buffer=io.BytesIO(); torch.save(obj,buffer); return buffer.getbuffer().nbytes
rf_test_batch=images[test_idx[:64]].reshape(min(64,len(test_idx)),-1)
for _ in range(5): rf.predict(rf_test_batch)
start=time.perf_counter()
for _ in range(40): rf.predict(rf_test_batch)
rf_ms=(time.perf_counter()-start)*1000/40
rf_pred=rf.predict(images[test_idx].reshape(len(test_idx),-1))
rows=[('Traditional RF',*metrics(rf_pred,labels[test_idx]),sum(t.tree_.node_count for t in rf.estimators_),rf_seconds,rf_ms,serialized_size(rf))]
for name,model,seconds in [('CNN scratch',scratch,scratch_seconds),('Task-transfer CNN',transfer,source_seconds+transfer_seconds)]:
    pred,truth=predictions(model,test_loader); rows.append((name,*metrics(pred,truth),sum(p.numel() for p in model.parameters()),seconds,model_latency(model,x[test_idx[:64]]),serialized_size(model.state_dict())))
print(f'{"Approach":20} {"Acc":>6} {"Prec":>6} {"Recall":>6} {"F1":>6} {"Params/nodes":>13} {"Train s":>9} {"Batch ms":>10} {"Size MiB":>9}')
for row in rows: print(f'{row[0]:20} {row[1]:6.3f} {row[2]:6.3f} {row[3]:6.3f} {row[4]:6.3f} {row[5]:13,d} {row[6]:9.2f} {row[7]:10.3f} {row[8]/1024**2:9.3f}')
```

## Complexity and trade-off report
| Approach | Complexity | Data need | Interpretability | Deployment considerations |
|---|---|---|---|---|
| Random Forest | low/moderate | modest on this simple data | feature importance, limited spatial explanation | CPU friendly; tree count affects size |
| CNN scratch | moderate | needs representative labels | saliency/feature maps available | compact task-specific model |
| Task transfer | more training steps | source data/task needed | CNN diagnostics | pretraining cost may amortize across tasks |
