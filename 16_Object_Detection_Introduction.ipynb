## Introduction to object detection

Object Detection is a computer vision task that identifies and locates multiple objects within an image or video. Unlike image classification, which assigns a label to an entire image, object detection predicts both the object category and its location using bounding boxes.

Enables real-time analysis of images and video streams.
Forms the foundation of applications such as autonomous driving, surveillance, and robotics.


In deep learning and computer vision, these terms describe different tasks that answering two fundamental questions: "What is in the image?" (classification) and "Where is it?" (localization).

Example:

Input: A traffic image

Output:

Car — Confidence: 0.95

Person — Confidence: 0.89

Bicycle — Confidence: 0.82

## Deep Dive into Each Concept


## 1. Image Classification

Image classification is the "Hello World" of computer vision. The neural network evaluates pixels across the entire image and assigns exactly one label from a predefined set of categories.

- How it works: Convolutional Neural Networks (CNNs) extract features and pass them to a final fully connected layer with a Softmax activation function to output a single probability vector.

- Limitation: It assumes there is only one dominant subject. It cannot tell you where the object is, nor can it handle multiple different objects cleanly


## 2. Multi-class Classification

This term refers strictly to the mathematical setup of the classification problem, regardless of whether you are classifying an entire image or a single cropped bounding box.

- The Distinction: Do not confuse this with multi-label classification (where an image can be a "dog" AND a "photo" at the same time). In multi-class classification, categories are mutually exclusive.

- Example: If your model looks at an image and must choose exactly one option from [Cat, Dog, Car, Person], it is performing multi-class classification.


## 3. Object Localization

Object localization focuses on the "Where" aspect. It is the task of identifying the exact physical coordinates and boundaries of an object within the frame.

- How it works: Instead of just predicting a discrete category, the network uses a regression head to predict four continuous numerical values: the center coordinates (x, y), width (w), and height (h) of a bounding box.

- Note on Object Detection: In deep learning architectures, Object Detection is simply the combination of both Image Classification and Object Localization executing simultaneously to find and label multiple items in a single scene.


## 4. Object Detection

Object detection is the combination of classification and localization applied to real-world complexity. It answers: "What objects are in this image, and where is each one?" Unlike localization, object detection can find multiple distinct objects belonging to multiple different classes all within the same frame.

- Example: In a street scene, the model simultaneously locates and labels three "Cars", two "Pedestrians", and one "Traffic Light" by drawing separate bounding boxes around every single one


## 5. Image Segmentation

While object detection stops at drawing rough rectangular boxes, image segmentation provides a granular, pixel-level understanding. It outlines the exact boundary of every object by classifying every single pixel in the image. This is divided into two types:

- Semantic Segmentation: Groups all pixels of the same category together (e.g., all car pixels are colored blue, all road pixels are colored green).

- Instance Segmentation: Tracks individual objects separately (e.g., Car #1 is blue, Car #2 is red).

- Example: In a medical scan, instead of just drawing a box around a tumor (detection), the model highlights the exact irregular shape of the tumor pixel-by-pixel so a surgeon knows its precise boundaries.

## Core concepts used in deep learning object detection:


## Bounding Boxes

- Definition: An imaginary rectangle that encloses an object of interest within an image.
- Purpose: It serves as the primary way a model pinpoints the physical location and scale of an object. Both human annotators (ground-truth) and models (predictions) use them.

```python
# Define bounding box coordinates

x_min = 100
y_min = 80
x_max = 300
y_max = 250

# Calculate bounding box dimensions
width = x_max - x_min
height = y_max - y_min

print("Bounding Box Coordinates:")
print("Left:", x_min)
print("Top:", y_min)
print("Right:", x_max)
print("Bottom:", y_max)

print("\nBounding Box Dimensions:")
print("Width:", width)
print("Height:", height)
```


## Coordinates

- Definition: The numeric spatial values that define the exact position and size of a bounding box.
- Common Formats:
	- [x_min, y_min, x_max, y_max]: Top-left and bottom-right corners.
	- [x_center, y_center, width, height]: Center point coordinates along with box dimensions.
    -  Can be measured raw in pixels or normalized to a 0–1 range relative to the image size.

```python
# Image dimensions
image_width = 640
image_height = 480

# Example point
x = 100
y = 80

print("Image Width:", image_width)
print("Image Height:", image_height)
print("Point Coordinates:", (x, y))

if 0 <= x < image_width and 0 <= y < image_height:
    print("The point is inside the image.")
else:
    print("The point is outside the image.")
```


## Class Labels

- Definition: The categorical identity assigned to a bounding box.
- Purpose: While coordinates handle where the object is, the class label identifies what the object is (e.g., "car", "person", "dog"). The model outputs a probability distribution across all target classes.


## Confidence Score

- Definition: A metric (usually between 0 and 1) reflecting the model's certainty that a bounding box actually contains an object, and that it belongs to the predicted class.
- Purpose: Acts as the first line of defense to filter out weak, background noise predictions before deeper post-processing occurs

```python
# Example dtection result

object_label = "Car"
confidence_score = 0.95
confidence_threshold = 0.70

print("Object:", object_label)
print("Confidence Score:", confidence_score)

if confidence_score >= confidence_threshold:
    print("Detection accepted.")
else:
    print("Detection filtered out.")
```

## Intersection over Union (IoU)

- Definition: A metric that evaluates the degree of spatial overlap between two bounding boxes.
- Formula: Calculated by dividing the area of overlap by the total combined area:
\(\text{IoU}=\frac{\text{Area\ of\ Intersection}}{\text{Area\ of\ Union}}\)
- Application: Used during model evaluation to see how closely a prediction matches the ground truth (where an IoU ≥ 0.5 is typically considered a correct match), and during NMS to identify duplicates.

## Non-Maximum Suppression (NMS)

- Definition: A vital post-processing algorithm used to clean up messy object detection outputs.
- Problem it Solves: Dense object detectors (like YOLO) often predict thousands of candidate boxes, causing several neighboring boxes to fire heavily on the exact same object.
